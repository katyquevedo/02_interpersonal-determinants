# Heatmap para Scoping review

```python
# Importar librerías
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("Listo!")
```

```python
# Datos de los estudios
data = {
    'Study app':      [10, 6, 4, 3, 1],
    'Social media':   [9, 3, 1, 1, 1],
    'Messaging':      [5, 2, 1, 0, 0],
    'Web-based':      [4, 2, 2, 0, 0],
    'Videocall':      [1, 0, 0, 0, 0]
}

index = [
    'Social support',
    'Social comparison',
    'Role modeling',
    'Social norms',
    'Parasocial processes'
]

df_heat = pd.DataFrame(data, index=index)
print(df_heat)
```

```python
from matplotlib.colors import LinearSegmentedColormap

# Crear gradiente personalizado de blanco a tu verde
cmap_custom = LinearSegmentedColormap.from_list(
    'custom_green', ['#ffffff', '#207065']
)
```

```python
# Crear mapa de calor
fig, ax = plt.subplots(figsize=(10, 5))

sns.heatmap(
    df_heat,
    annot=True,        # muestra los números dentro de cada celda
    fmt='d',           # formato entero (sin decimales)
    cmap=cmap_custom,      # escala de colores azules
    linewidths=0.5,    # líneas entre celdas
    linecolor='white',
    ax=ax
)

#ax.set_title('Number of studies per interpersonal determinant and delivery channel',
             #fontsize=12, pad=15)
ax.set_xlabel('Digital delivery channel', fontsize=10)
ax.set_ylabel('Interpersonal determinant', fontsize=10)
ax.tick_params(axis='x', rotation=0)
ax.tick_params(axis='y', rotation=0)

plt.tight_layout()
plt.savefig('../outputs/heatmap_determinants.png', dpi=300, bbox_inches='tight')
plt.show()
```
