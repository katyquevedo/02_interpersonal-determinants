# Heatmap para Scoping review

```python
# Import libraries
import pandas as pd              # Organizar data into dataframes
import matplotlib.pyplot as plt  # Create figures
import seaborn as sns            # For publication-quality graphics
from matplotlib.colors import LinearSegmentedColormap  # Custom color gradients


print("Libraries loaded successfully!")
```

```python
# Input data
    # Rows: Interpersonal determinants
    # Columns: Delivery platforms
    # Values: Number of papers

data = {
    'Study apps':      [12, 6, 4, 3, 1],
    'Social media':   [10, 3, 1, 1, 1],
    'Messaging':      [6, 2, 1, 0, 0],
    'Website':      [5, 2, 2, 0, 0],
    'Videocall':      [2, 0, 0, 0, 0]
}

index = [
    'Social support',
    'Social comparison',
    'Role modeling',
    'Social norms',
    'Parasocial processes'
]
```

```python
# Create DataFrame
df_heat = pd.DataFrame(data, index=index)

print(df_heat)
```

```python
# Create personalized color gradient (white to INDEEP green)
colmap_green = LinearSegmentedColormap.from_list(
    'custom_green', ['#ffffff', '#207065']
)
```

```python
# Create heatmap

# Create a figure (canvas) and axis (plot area) for the heatmap
fig, ax = plt.subplots(figsize=(8, 5))  

# Draw the heatmap
sns.heatmap(
    df_heat,
    annot=True,                 # Display values inside each cell
    fmt='d',                    # Integer format (no decimals)
    annot_kws={"size":14},      # Size of numbers inside cells
    cmap=colmap_green,          # Custom green color palette
    linewidths=0.5,             # Add white borders between cells
    linecolor='white',
    cbar_kws={
        "shrink": 0.8,          # Make the colorbar 80% of its original height
        "aspect": 30},          # Make the color bar thinner
    ax=ax
)

# Figure title (decided to exclude for now)
#ax.set_title('Number of studies per interpersonal determinant and delivery channel',
             #fontsize=12, pad=15)

# Label de x and y axis
ax.set_xlabel('Digital delivery channel', fontsize=11)
ax.set_ylabel('Interpersonal determinant', fontsize=11)

# Keep axis labels horizontal
ax.tick_params(axis='x', rotation=0, labelsize=10)
ax.tick_params(axis='y', rotation=0, labelsize=10)

# Adjust spacing to avoid overlapping labels
plt.tight_layout()

# Save the figure
plt.savefig('../outputs/heatmap_determinants.png', dpi=300, bbox_inches='tight')

# Display the figure in notebook
plt.show()
```
